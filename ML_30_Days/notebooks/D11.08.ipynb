### 1.1 linear regression
1.1.1假设我们有一些数据$x1,...,xn\in \mathbb{R}$我们的目标是找到一个常数b，是的最小化$$\sum_{i}(xi-b)^{2}$$
A.找到最优值b的解析解：
answer：
考虑这个数据是赋范线性空间$\mathbb{R}^{n}$的一个向量，而$\vec{b}$是每个坐标都相等的向量，这些向量构成的集合是通过原点的一条直线。而我们要最小化的就是这个直线上的点到数据的距离的平方。
因此我们只需要取这个数据点到直线的距离即可
考虑直线上的点：$$\vec{x}=(x,x,...,x)=x(1,1,...,1)\in\mathbb{R}^{n}$$
这个点到数据点的向量是：$$\vec{v}-\vec{x}=(x_1-x,x_2-x,...,x_n-x)$$
这个向量和直线的方向向量垂直：$$\implies (\vec{v}-\vec{x})*\vec1=0$$
$$\implies \sum_{i=1}^{n}x_{i}=\sum_{i=1}^{n}x=nx$$
$$\implies x= \frac{1}{n}\sum_{i=1}^{n}x_{i}$$

1.1.2  推导出使用平方误差的线性回归优化问题的解析解，为了简化问题，可以忽略偏置b（我们可以通过像$\mathbf{X}$添加所有值为1的一列来做到这一点）  

A:用矩阵和向量表示法写出优化问题  

Answer:  

考虑我们的问题有n个实特征$f_1,f_2,...,f_n$,有m个数据点$(x_1,y_1),...,(x_m,y_m)\in\mathbb{R}^{n+1}$,其中y是对应的标签
如果我们的拟合线性函数是$$\hat{y}=\mathbf{\omega}*\vec{x}$$
那么损失函数是：$$loss(\mathbf{\omega})=\left(\frac{1}{n}\sum_{i=1}^{m}(\mathbf{\omega}*\vec{x_{i}}-y_{i})^{2}\right)^{\frac{1}{2}}$$
  
B: 计算损失对$\mathbf{\omega}$的梯度

Answer:

$$
\begin{align}
\nabla{loss(\mathbf{\omega})}&=\nabla{\left(\frac{1}{n}\sum_{i=1}^{m}(\mathbf{\omega}*\vec{x_{i}}-y_{i})^{2}\right)^{\frac{1}{2}}} \\
&=\frac{\sum_{i=1}^{m}(\mathbf{\omega}x_{i}-y_{i})x_{i}}{\sqrt{\frac{1}{n}\sum_{i=1}^{m}(\mathbb{\omega}x_{i}-y_{i})^{2}}}
\end{align}$$

C:通过将梯度设为0，求解矩阵方程来找到解析解

Answer:

$$\nabla{loss{\omega}}=0$$
$$\implies \mathbf{\omega}=(\mathbf{X}^{T}\mathbf{X})^{-1}\mathbf{X}^{T}\mathbf{y}$$

D:什么时候可能使用随机梯度下降更好？这种方法何时会失效？

Answer：

这种做法要求由数据X拼成的矩阵是可逆的，在不可逆的时候只能使用Moore广义逆来解决。


1.1.3 假定控制附加噪声$\varepsilon$的噪声模型是指数分布，也就是说：$p(\varepsilon)=\frac{1}{2}\exp{-|\varepsilon|}$

A:写出模型$-log P(y|\mathbf{X})$下数据的负对数似然

answer:

我们假设我们的预估模型是线性的，然后存在一些噪声，这些噪声的分布遵循指数分布：
$$y=\mathbf{w}^{T}\mathbf{x}+b+\varepsilon$$
因此，我们可以写出通过给定的x观测到待定的y的似然：
$$P(y|\mathbf{x})=\frac{1}{2}\exp(-|y-\mathbf{w}^{T}\mathbf{x}-b|)$$
根据极大似然估计法：
参数$\mathbf{w},b$的最优值是使得整个数据集的似然最大的值：
$$P(\vec{y}|\mathbf{X})=\prod_{i=1}^{m}p(y_{i}|\mathbf{x}_{i})$$
那么负对数似然就是：
$$\begin{align}
-logP(\vec{y}|\mathbf{X})&=-log\prod_{i=1}^{m}p(y_{i}|\mathbf{x}_{i})
\\ &=\sum_{i=1}^{m}-logp(y_{i}|\mathbf{x}_{i})
\\ &=\sum_{i=1}^{m}-log \frac{1}{2}\exp(-|y_{i}-\mathbf{w}^{T}\mathbf{x_{i}}-b|)
\\ &=\sum_{i=1}^{m}|y_{i}-\mathbf{w}^{T}\mathbf{x_{i}}-b|
\end{align}$$

B:请试着写出解析解

answer：

我们想要负对数似然最大，正常的思路就是把负对数似然看成关于参数$\omega$的一个函数，我们求关于$\omega$的梯度，算出驻点。
但是这里是绝对值的求和，在每个数据点和预测点残差为0的地方函数都不可导
退而求其次：如果我们在不变号的区间上看能不能找到？
我们需要的是$$\begin{cases}\sum_{i=1}^{m}sgn(r_{i})\mathbf{x}_{i}=0
\\
\\\sum_{i=1}^{m}sgn(r_{i})=0\end{cases}$$
如果我们要知道残差的正负，就需要参数
我们要求出参数，就需要残差的正负
所以没有解析解
